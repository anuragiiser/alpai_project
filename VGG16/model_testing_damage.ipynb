```python
import numpy as np
import os

X_test = np.load( "/home/sombit-ng-nonadm/vaibhav_intern_pg/BTP+AIproject/task2_damage_state_1/task2_X_test.npy")
y_test = np.load("/home/sombit-ng-nonadm/vaibhav_intern_pg/BTP+AIproject/task2_damage_state_1/task2_y_test.npy")
print(X_test.shape)
print(y_test.shape)
```

```python
from tensorflow.keras.models import load_model

# Load model without compiling (since you're only using it for prediction)
model = load_model(
    "/home/sombit-ng-nonadm/vaibhav_intern_pg/BTP+AIproject/VGG16/Models/retrained_model_damage_medium.h5",
    compile=False
)
```

```python
predictions = model.predict(X_test)
```

```python
predicted_classes = np.argmax(predictions, axis=1)

print(predicted_classes[:100])
```

```python
y_test_labels = np.argmax(y_test, axis=1)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, classification_report

# Convert model predictions to class labels
y_pred_labels = np.argmax(predictions, axis=1)

# Now compute accuracy
test_accuracy = accuracy_score(y_test_labels, y_pred_labels)
report= classification_report(y_test_labels, y_pred_labels, target_names=['Class 0', 'Class 1'], output_dict=True)
print(report)
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"macro avg f1-score: {report['macro avg']['f1-score']:.4f}")
```
